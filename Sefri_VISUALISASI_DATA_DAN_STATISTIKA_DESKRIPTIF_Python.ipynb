```python
# Membuat tabel distribusi frekuensi Grade
datafreq = pd.crosstab(
    index=data["Grade"],
    columns="Frekuensi"
)

# Memberikan nama pada index
datafreq.index.name = "Grade"

# Menampilkan tabel
display(datafreq)
```

```python
import matplotlib.pyplot as plt

datafreq.plot(
    kind="line",
    marker="o",
    figsize=(8,5),
    linewidth=2
)

plt.title("Distribusi Frekuensi Grade Mahasiswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.grid(True, alpha=0.3)
plt.legend(["Frekuensi"])
plt.tight_layout()
plt.show()
```

```python
datafreq.plot(
    kind="bar",
    figsize=(8,5),
    legend=False
)

plt.title("Distribusi Frekuensi Grade Mahasiswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
datafreq.plot(
    kind="pie",
    y="Frekuensi",
    autopct="%1.1f%%",
    figsize=(7,7),
    legend=False,
    startangle=90
)

plt.title("Persentase Grade Mahasiswa")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Mengambil data Final Score
dt = pd.to_numeric(data["Final Score"], errors="coerce").dropna()

# Membuat statistika deskriptif
hasil_statistik = pd.DataFrame({
    "Statistik": [
        "Jumlah Data",
        "Mean",
        "Median",
        "Modus",
        "Minimum",
        "Q1",
        "Q2",
        "Q3",
        "Maximum",
        "Range",
        "Variance",
        "Standard Deviation",
        "Standard Error",
        "Skewness",
        "Kurtosis"
    ],
    "Nilai": [
        dt.count(),
        dt.mean(),
        dt.median(),
        dt.mode().iloc[0],
        dt.min(),
        dt.quantile(0.25),
        dt.quantile(0.50),
        dt.quantile(0.75),
        dt.max(),
        dt.max() - dt.min(),
        dt.var(),
        dt.std(),
        dt.sem(),
        dt.skew(),
        dt.kurt()
    ]
})

# Membulatkan hasil
hasil_statistik["Nilai"] = hasil_statistik["Nilai"].round(2)

display(hasil_statistik)
```
